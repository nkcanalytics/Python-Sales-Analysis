```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    "Order_ID": range(101, 116),
    
    "Product": [
        "Laptop", "Mobile", "Tablet", "Laptop", "Mobile",
        "Headphones", "Tablet", "Laptop", "Mobile", "Headphones",
        "Laptop", "Tablet", "Mobile", "Headphones", "Laptop" ],
    
    "Region": [
        "South", "North", "East", "West", "South",
        "North", "East", "South", "West", "North",
        "East", "West", "South", "East", "North" ],
    
    "Quantity": [2, 3, 1, 2, 4, 5, 2, 1, 3, 4, 2, 3, 2, 1, 3],
    "Unit_Price": [50000, 20000, 15000, 50000, 20000,
                   2000, 15000, 50000, 20000, 2000,
                   50000, 15000, 20000, 2000, 50000]
}

df = pd.DataFrame(data)
```

```python
df =pd.DataFrame(data)
```

```python
df["Revenue"] = df["Quantity"] * df["Unit_Price"]
```

```python
df.head()
```

```python
total_revenue = df["Revenue"].sum()
```

```python
product_sales = df.groupby("Product")["Revenue"].sum()
```

```python
region_sales = df.groupby("Region")["Revenue"].sum()
```

```python
plt.figure(figsize=(8,5))

product_sales.plot(kind="bar")

plt.title("Product Wise Sales")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

region_sales.plot(kind="bar")

plt.title("Region Wise Sales")
plt.xlabel("Region")
plt.ylabel("Revenue")

plt.show()
```

```python
product_sales.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7,7)
)

plt.title("Product Sales Distribution")
plt.ylabel("")

plt.show()
```

```python
best_product = product_sales.idxmax()

print("Best Performing Product:", best_product)
```

```python
best_region = region_sales.idxmax()

print("Best Performing Region:", best_region)
```

```python
average_revenue = df["Revenue"].mean()

print("Average Order Revenue:", average_revenue)
```

```python
total_quantity = df["Quantity"].sum()

print("Total Quantity Sold:", total_quantity)
```
